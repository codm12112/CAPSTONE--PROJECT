```python
import os

# Path to the directory containing the YOLO detection txt files
labels_dir = '/content'

# Iterate over each .txt file in the directory
for filename in os.listdir(labels_dir):
    if filename.endswith('.txt'):
        file_path = os.path.join(labels_dir, filename)
        with open(file_path, 'r') as file:
            lines = file.readlines()

        # Filter out lines that contain '(no detections)'
        cleaned_lines = [line for line in lines if '(no detections)' not in line.lower()]

        # Write back the cleaned lines to the same file
        with open(file_path, 'w') as file:
            file.writelines(cleaned_lines)

print("Cleaned all text files successfully.")
```

```python
import subprocess
import re
import os
import csv

# Define your class names (in the same order as YOLO training)
class_names = ['Hello', 'Yes', 'No']  # Update this list with all your class names

# Run YOLOv8 and capture output
result = subprocess.run(
    [
        'yolo',
        'task=detect',
        'mode=predict',
        'model=/content/runs/detect/train/weights/best.pt',
        'conf=0.25',
        'source=/content/test/capstone (1).mp4',
        'save=True',
        'save_txt=True'
    ],
    stdout=subprocess.PIPE,
    stderr=subprocess.STDOUT,
    text=True
)

# Regex to extract inference times like '43.9ms'
time_pattern = re.compile(r'(\d+\.\d+)ms')

# Find the latest 'predict' folder
base_dir = "/content/runs/detect"
predict_folders = sorted([f.path for f in os.scandir(base_dir) if f.is_dir() and 'predict' in f.name])
latest_predict = predict_folders[-1]
labels_dir = os.path.join(latest_predict, "labels")

# List label files (each frame has one)
label_files = sorted([
    os.path.join(labels_dir, f) for f in os.listdir(labels_dir)
    if f.endswith('.txt')
])

# Prepare CSV data
output_rows = []
frame_index = 0

# Parse YOLO output and match with label files
for line in result.stdout.splitlines():
    time_match = time_pattern.search(line)
    if time_match and frame_index < len(label_files):
        time_taken = time_match.group(1)

        label_path = label_files[frame_index]
        with open(label_path, 'r') as f:
            detections = f.readlines()

        for det in detections:
            parts = det.strip().split()
            if not parts:
                continue
            class_id = int(parts[0])
            label = class_names[class_id] if class_id < len(class_names) else f"Class{class_id}"
            output_rows.append([frame_index + 1, label, time_taken])

        frame_index += 1

# Write to CSV
csv_path = "/content/all_predictions.csv"
with open(csv_path, mode='w', newline='') as file:
    writer = csv.writer(file)
    writer.writerow(["Frame", "Label", "Time(ms)"])  # Header
    writer.writerows(output_rows)

print(f"✅ All predictions saved to: {csv_path}")
```

```python
import csv
import re

input_path = "/content/output.txt"
output_path = "/content/predictions_only.csv"

rows = []

with open(input_path, 'r') as infile:
    for line in infile:
        line = line.strip()

        # Check for 'no detections'
        if "no detections" in line.lower():
            rows.append(["no detections"])
        else:
            # Try to extract prediction label like: '1 Hello, 43.9ms'
            match = re.search(r'\d+\s+([A-Za-z_]+),\s+\d+\.?\d*ms', line)
            if match:
                label = match.group(1)
                rows.append([label])
            else:
                rows.append(["(unknown)"])

# Write to CSV with only the label column
with open(output_path, 'w', newline='') as csvfile:
    writer = csv.writer(csvfile)
    writer.writerow(['Label'])  # Header
    writer.writerows(rows)

print(f"✅ Predictions saved to: {output_path}")
```
